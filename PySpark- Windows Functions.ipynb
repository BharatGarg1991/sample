```python
data = spark.read.csv("C:\\Users\\bgarg\\Downloads\\sales_info.csv",header=True,inferSchema=True)
```

```python
data.show()
```

```python
data=data.withColumn("prvs_sales",lag("T_Sales").over(Window.orderBy(asc("C_Year")) ))
```

```python
data.show()
```

```python

data=data.withColumn("YOY",((col("T_Sales")-col("prvs_sales"))/col("prvs_sales")*100))
```

```python
data.show()
```
